# Training calibrated models with XGBoost

```python
# Cell 1
import os, joblib, numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay, roc_curve, auc, precision_recall_curve, average_precision_score, roc_auc_score
from xgboost import XGBClassifier
from sklearn.calibration import CalibratedClassifierCV

OUTPUT_DIR = "model_outputs_calibrated"
os.makedirs(OUTPUT_DIR, exist_ok=True)
pd.options.display.max_columns = 300
```

```python
import numpy as np
import pandas as pd

def first_existing_series(df, candidates):
    """Return first Series in df among candidates list, or None if none present."""
    for c in candidates:
        if c in df.columns:
            return df[c]
    return None

def add_derived_features(df):
    df = df.copy()
    # convert numeric-like object cols to numeric where possible
    for c in df.columns:
        if df[c].dtype == object:
            df[c] = pd.to_numeric(df[c], errors='ignore')

    # ensure forward/backward packet counts exist (create if not present)
    if 'total_fwd_packets' not in df.columns:
        df['total_fwd_packets'] = 0
    if 'total_backward_packets' not in df.columns:
        # some datasets use 'total_backward_packets' naming variants
        if 'total_backwards_packets' in df.columns:  # example typo fallback (if any)
            df['total_backward_packets'] = df['total_backwards_packets']
        else:
            df['total_backward_packets'] = 0

    # total packets
    df['total_packets'] = df['total_fwd_packets'].fillna(0) + df['total_backward_packets'].fillna(0)

    # total bytes: pick first existing column among common candidates
    fwd_candidates = ['fwd_packets_length_total', 'total_length_of_fwd_packets', 'subflow_fwd_bytes', 'Subflow Fwd Bytes', 'subflow_fwd_bytes']
    bwd_candidates = ['bwd_packets_length_total', 'total_length_of_bwd_packets', 'subflow_bwd_bytes', 'Subflow Bwd Bytes', 'subflow_bwd_bytes']

    fwd_bytes_series = first_existing_series(df, fwd_candidates)
    bwd_bytes_series = first_existing_series(df, bwd_candidates)

    if fwd_bytes_series is not None:
        df['total_fwd_bytes'] = fwd_bytes_series.fillna(0)
    else:
        df['total_fwd_bytes'] = 0

    if bwd_bytes_series is not None:
        df['total_bwd_bytes'] = bwd_bytes_series.fillna(0)
    else:
        df['total_bwd_bytes'] = 0

    df['total_bytes'] = df['total_fwd_bytes'] + df['total_bwd_bytes']

    # duration safe (ensure numeric and positive + tiny epsilon)
    if 'flow_duration' in df.columns:
        df['flow_duration'] = pd.to_numeric(df['flow_duration'], errors='coerce').fillna(0).clip(lower=0) + 1e-6
    else:
        df['flow_duration'] = 1.0

    # derived features (guards for zero division)
    df['packet_rate'] = df['total_packets'] / df['flow_duration']
    df['byte_rate'] = df['total_bytes'] / df['flow_duration']
    df['mean_packet_size'] = (df['total_bytes'] / df['total_packets'].replace({0: np.nan})).fillna(0)
    df['fwd_ratio'] = (df['total_fwd_packets'] / df['total_packets'].replace({0: np.nan})).fillna(0)

    # iat_range if columns exist (safe)
    if 'flow_iat_max' in df.columns and 'flow_iat_min' in df.columns:
        df['iat_range'] = (pd.to_numeric(df['flow_iat_max'], errors='coerce').fillna(0) -
                           pd.to_numeric(df['flow_iat_min'], errors='coerce').fillna(0)).clip(lower=0)
    else:
        df['iat_range'] = 0.0

    # log transforms (handle negatives/NaNs by log1p on non-negative)
    df['log_packet_rate'] = np.log1p(df['packet_rate'].clip(lower=0))
    df['log_byte_rate'] = np.log1p(df['byte_rate'].clip(lower=0))
    df['log_total_bytes'] = np.log1p(df['total_bytes'].clip(lower=0))
    df['log_total_packets'] = np.log1p(df['total_packets'].clip(lower=0))

    return df

# Apply corrected function to your dfs
cic_proc = add_derived_features(cic_proc)
cicids_proc = add_derived_features(cicids_proc)

print("Derived features added without errors.")
print("cic_proc shape:", cic_proc.shape)
print("cicids_proc shape:", cicids_proc.shape)
```

```python
# mix fraction: change if needed
frac = 0.03

cicids_attacks = cicids_proc[cicids_proc['label'] == 1]
mix_n = max(1, int(len(cicids_attacks) * frac))
cicids_sample = cicids_attacks.sample(n=mix_n, random_state=42)

train_mix = pd.concat([cic_proc, cicids_sample], ignore_index=True).sample(frac=1.0, random_state=42).reset_index(drop=True)
print("train_mix shape:", train_mix.shape, "added attack rows:", mix_n)

# select numeric common features that exist in both train_mix and cicids_proc
num_train = train_mix.select_dtypes(include=[np.number]).columns.tolist()
num_test = cicids_proc.select_dtypes(include=[np.number]).columns.tolist()
if 'label' in num_train: num_train.remove('label')
if 'label' in num_test: num_test.remove('label')

features = [c for c in num_train if c in num_test]
print("Using numeric features count:", len(features))
print("Sample features:", features[:20])
```

### 🧠 Model Evaluation & Optimization Plan

**Đánh giá mô hình hiện tại:**
- **Ưu điểm**: Sử dụng `tree_method='hist'` tối ưu tốc độ. Có chiến lược Calibration (quan trọng cho Imbalanced Data).
- **Nhược điểm**: `max_depth=10` hơi cao (dễ Overfitting). `n_estimators=2000` fix cứng. Chưa có bước tìm tham số tối ưu.
- **Sửa lỗi**: Đã cập nhật `early_stopping_rounds` vào constructor để tương thích XGBoost v1.6+.

**Tối ưu hóa:**
- Giảm `max_depth` xuống 8, `learning_rate` xuống 0.03 để mô hình học chậm và chắc hơn.
- Thêm đoạn code `RandomizedSearchCV` bên dưới để bạn có thể chạy tìm bộ tham số tốt nhất tự động.

```python
# === TRAIN/VAL SPLIT + scale_pos_weight ===
from sklearn.model_selection import train_test_split

# Use the mixed training data and the already-computed feature list
X = train_mix[features].copy()
y = train_mix['label'].astype(int).copy()

# Stratified split to preserve class ratio
X_tr, X_val, y_tr, y_val = train_test_split(
    X, y,
    test_size=0.10,
    stratify=y,
    random_state=42
)

# Compute scale_pos_weight on the TRAIN split (recommended for XGBoost)
pos = int((y_tr == 1).sum())
neg = int((y_tr == 0).sum())
scale_pos_weight = float(neg / pos) if pos > 0 else 1.0

print("="*80)
print("📊 TRAIN SPLIT STATS")
print("="*80)
print(f"Train size: {len(y_tr)} | Val size: {len(y_val)}")
print(f"Class counts (train): benign={neg:,}, attack={pos:,}")
print(f"scale_pos_weight = {scale_pos_weight:.4f}  # (neg/pos)")
```

```python
# Fixed & Improved Configuration
model = XGBClassifier(
    n_estimators=10000, # Increased for early stopping flexibility
    max_depth=8,        # Slightly reduced to prevent overfitting
    learning_rate=0.03, # Lower learning rate for better generalization
    subsample=0.7,      # Lower subsample for regularization
    colsample_bytree=0.7,
    tree_method="hist",
    n_jobs=-1,
    random_state=42,
    use_label_encoder=False,
    eval_metric="auc",
    scale_pos_weight=scale_pos_weight,
    early_stopping_rounds=100 # Moved to constructor (XGBoost >= 1.6)
)

model.fit(
    X_tr, y_tr,
    eval_set=[(X_val, y_val)],
    verbose=100
)

# quick eval on CICIDS2017
X_test = cicids_proc[features]
y_test = cicids_proc['label'].astype(int)
y_prob_raw = model.predict_proba(X_test)[:,1]
print("Before calibration:")
print(classification_report(y_test, (y_prob_raw>=0.5).astype(int), digits=4))
print("ROC AUC before:", roc_auc_score(y_test, y_prob_raw))
```

```python
# === OPTIMIZATION SECTION (Optional) ===
# Run this cell to find better parameters using RandomizedSearchCV
from sklearn.model_selection import RandomizedSearchCV

SEARCH = False  # Set to True to run optimization (Time consuming!)

if SEARCH:
    print("Running Hyperparameter Optimization...")
    
    # Define parameter grid
    param_dist = {
        'n_estimators': [1000, 3000, 5000],
        'max_depth': [3, 6, 8, 10],
        'learning_rate': [0.01, 0.03, 0.05, 0.1],
        'subsample': [0.6, 0.7, 0.8, 0.9],
        'colsample_bytree': [0.6, 0.7, 0.8],
        'gamma': [0, 0.1, 0.5, 1],
        'min_child_weight': [1, 3, 5],
        'scale_pos_weight': [scale_pos_weight, scale_pos_weight * 1.2, scale_pos_weight * 0.8]
    }
    
    xgb_base = XGBClassifier(
        tree_method='hist', 
        eval_metric='auc', 
        early_stopping_rounds=50, # Note: For CV, early_stopping might require fit_params or manual handling. 
        # Here we skip early_stopping in CV loop for simplicity or pass it via fit_params if supported
        use_label_encoder=False,
        random_state=42,
        n_jobs=-1
    )

    # Randomized Search
    # Note: early_stopping_rounds in CV is tricky because we need eval_set. 
    # Usually we tune without early_stopping or split internal CV.
    # For simplicity, we just search with fixed n_estimators and no early stopping in the loop,
    # OR we use a smaller n_estimators for search.
    
    random_search = RandomizedSearchCV(
        estimator=xgb_base,
        param_distributions=param_dist,
        n_iter=20, # Number of parameter settings to try
        scoring='roc_auc',
        cv=3,
        verbose=2,
        random_state=42,
        n_jobs=-1
    )
    
    # Fit search (using only X_tr to prevent leakage, though X_tr is already a split)
    random_search.fit(X_tr, y_tr)
    
    print("Best Parameters found:", random_search.best_params_)
    print("Best ROC AUC:", random_search.best_score_)
    
    # Re-train model with best params
    model = random_search.best_estimator_
    # Ensure early stopping setup for final fit if needed
    # model.set_params(early_stopping_rounds=100, n_estimators=10000)
    model.fit(X_tr, y_tr, eval_set=[(X_val, y_val)], verbose=100)
```

```python
from sklearn.calibration import CalibratedClassifierCV
from copy import deepcopy
from sklearn.metrics import classification_report, roc_auc_score, brier_score_loss

# clone model to avoid modifying original
model_clone = deepcopy(model)

# !!! CRITICAL FIX: Disable early_stopping_rounds for Calibration
# CalibratedClassifierCV uses internal Cross Validation and does not pass 'eval_set'
# If early_stopping_rounds is set, XGBoost crashes demanding eval_set.
model_clone.set_params(early_stopping_rounds=None)

# CalibratedClassifierCV will cross-validate internally
calib_cv = CalibratedClassifierCV(estimator=model_clone, method='sigmoid', cv=3)
calib_cv.fit(X_tr, y_tr)

proba_test_cvcal = calib_cv.predict_proba(X_test)[:,1]

print("After CalibratedClassifierCV (refit) --- threshold=0.5:")
print(classification_report(y_test, (proba_test_cvcal>=0.5).astype(int), digits=4))
print("ROC AUC:", roc_auc_score(y_test, proba_test_cvcal), "Brier:", brier_score_loss(y_test, proba_test_cvcal))
```

```python
# === THRESHOLD OPTIMIZATION ===
# Find the best threshold to reduce False Positives while keeping High Recall
import numpy as np

print("Searching for optimal threshold...")
thresholds = np.arange(0.5, 0.99, 0.01)

best_t = 0.5
best_score = -1

print(f"{'Threshold':<10} | {'FP Rate (Normal block)':<20} | {'FN Rate (Attack miss)':<20} | {'F1-Score':<10}")
print("-"*80)

for t in thresholds:
    y_pred = (proba_test_cvcal >= t).astype(int)
    
    # Confusion Matrix components
    # 0 = Normal, 1 = Attack
    # FP: Normal predicted as Attack (We want to MINIMIZE this)
    fp = ((y_pred == 1) & (y_test == 0)).sum()
    tn = ((y_pred == 0) & (y_test == 0)).sum()
    fp_rate = fp / (fp + tn) if (fp + tn) > 0 else 0
    
    # FN: Attack predicted as Normal (We want to MINIMIZE this too, but FP is more annoying for users)
    fn = ((y_pred == 0) & (y_test == 1)).sum()
    tp = ((y_pred == 1) & (y_test == 1)).sum()
    fn_rate = fn / (fn + tp) if (fn + tp) > 0 else 0
    
    f1 = 2 * tp / (2 * tp + fp + fn)
    
    # Simple heuristic: heavily penalize FP
    # Score = F1 - 2 * FP_Rate
    score = f1 - 2 * fp_rate
    
    if fp_rate < 0.05: # Only show interesting ones
        print(f"{t:.2f}       | {fp_rate:.4f}               | {fn_rate:.4f}               | {f1:.4f}")
    
    if score > best_score:
        best_score = score
        best_t = t

print("-"*80)
print(f"Recommended Threshold: {best_t:.2f}")

# Evaluate with recommended threshold
print(f"\nFINAL EVALUATION at Threshold {best_t:.2f}:")
print(classification_report(y_test, (proba_test_cvcal >= best_t).astype(int), digits=4))
```
